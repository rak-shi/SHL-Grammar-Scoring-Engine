```python
# Import librosa for audio and speech features
import librosa

# Import NumPy for numerical calculations
import numpy as np

# Import pandas for creating feature tables
import pandas as pd

# Import os for file paths
import os

# Import tqdm for a progress bar
from tqdm.auto import tqdm


# Define a function to extract acoustic and prosodic features
def extract_acoustic_features(audio_path):

    # Load the audio at 16 kHz
    y, sr = librosa.load(
        audio_path,
        sr=16000,
        mono=True
    )

    # Protect against empty audio files
    if len(y) == 0:
        return np.zeros(25, dtype=np.float32)

    # Calculate audio duration in seconds
    duration = len(y) / sr

    # Calculate RMS energy
    rms = librosa.feature.rms(y=y)[0]

    # Calculate zero-crossing rate
    zcr = librosa.feature.zero_crossing_rate(y)[0]

    # Calculate spectral centroid
    centroid = librosa.feature.spectral_centroid(
        y=y,
        sr=sr
    )[0]

    # Calculate spectral bandwidth
    bandwidth = librosa.feature.spectral_bandwidth(
        y=y,
        sr=sr
    )[0]

    # Calculate spectral rolloff
    rolloff = librosa.feature.spectral_rolloff(
        y=y,
        sr=sr
    )[0]

    # Calculate pitch using librosa's YIN algorithm
    f0 = librosa.yin(
        y,
        fmin=70,
        fmax=400,
        sr=sr
    )

    # Keep only valid pitch values
    valid_f0 = f0[np.isfinite(f0)]

    # Calculate onset strength
    onset = librosa.onset.onset_strength(
        y=y,
        sr=sr
    )

    # Calculate short-time energy changes
    energy_change = np.diff(rms)

    # Calculate approximate silence proportion
    silence_ratio = np.mean(rms < (np.max(rms) * 0.05))

    # Calculate speech activity ratio
    speech_ratio = 1.0 - silence_ratio

    # Calculate estimated speaking activity from onset rate
    onset_rate = len(
        librosa.onset.onset_detect(
            y=y,
            sr=sr
        )
    ) / max(duration, 1e-6)

    # Calculate pitch statistics
    if len(valid_f0) > 0:

        # Mean pitch
        pitch_mean = np.mean(valid_f0)

        # Standard deviation of pitch
        pitch_std = np.std(valid_f0)

        # Minimum pitch
        pitch_min = np.min(valid_f0)

        # Maximum pitch
        pitch_max = np.max(valid_f0)

    else:

        # Use zeros if pitch could not be detected
        pitch_mean = 0.0
        pitch_std = 0.0
        pitch_min = 0.0
        pitch_max = 0.0

    # Build the acoustic feature vector
    features = [

        # Duration
        duration,

        # RMS statistics
        np.mean(rms),
        np.std(rms),
        np.min(rms),
        np.max(rms),

        # ZCR statistics
        np.mean(zcr),
        np.std(zcr),

        # Spectral centroid statistics
        np.mean(centroid),
        np.std(centroid),

        # Spectral bandwidth statistics
        np.mean(bandwidth),
        np.std(bandwidth),

        # Spectral rolloff statistics
        np.mean(rolloff),
        np.std(rolloff),

        # Pitch statistics
        pitch_mean,
        pitch_std,
        pitch_min,
        pitch_max,

        # Onset statistics
        np.mean(onset),
        np.std(onset),

        # Energy change statistics
        np.mean(energy_change) if len(energy_change) else 0.0,
        np.std(energy_change) if len(energy_change) else 0.0,

        # Silence/speech statistics
        silence_ratio,
        speech_ratio,

        # Estimated onset rate
        onset_rate,

        # Number of detected pitch frames
        len(valid_f0)
    ]

    # Return the feature vector as float32
    return np.asarray(
        features,
        dtype=np.float32
    )


# Create a list for training acoustic features
train_acoustic = []

# Extract features from every training audio file
for filename in tqdm(
    train_df["filename"].tolist(),
    desc="Extracting train acoustic features"
):

    # Build the audio path
    audio_path = os.path.join(
        TRAIN_DIR,
        filename
    )

    # Extract acoustic features
    features = extract_acoustic_features(
        audio_path
    )

    # Store the features
    train_acoustic.append(features)


# Create a list for test acoustic features
test_acoustic = []

# Extract features from every test audio file
for filename in tqdm(
    test_df["filename"].tolist(),
    desc="Extracting test acoustic features"
):

    # Build the audio path
    audio_path = os.path.join(
        TEST_DIR,
        filename
    )

    # Extract acoustic features
    features = extract_acoustic_features(
        audio_path
    )

    # Store the features
    test_acoustic.append(features)


# Convert training features into a NumPy matrix
train_acoustic = np.vstack(train_acoustic)

# Convert test features into a NumPy matrix
test_acoustic = np.vstack(test_acoustic)


# Print the training feature shape
print("Train acoustic shape:", train_acoustic.shape)

# Print the test feature shape
print("Test acoustic shape:", test_acoustic.shape)
```

```python
# Combine Wav2Vec2 embeddings with acoustic features for training
X_train_raw = np.hstack([
    train_embeddings,
    train_acoustic
])

# Combine Wav2Vec2 embeddings with acoustic features for testing
X_test_raw = np.hstack([
    test_embeddings,
    test_acoustic
])

# Extract the target grammar scores
y = train_df["label"].values.astype(np.float32)

# Print the combined training shape
print("Combined train shape:", X_train_raw.shape)

# Print the combined test shape
print("Combined test shape:", X_test_raw.shape)

# Print the target shape
print("Target shape:", y.shape)

# Check for missing values in the training features
print(
    "Training NaN count:",
    np.isnan(X_train_raw).sum()
)

# Check for missing values in the test features
print(
    "Test NaN count:",
    np.isnan(X_test_raw).sum()
)
```

```python
# Import StandardScaler for feature normalization
from sklearn.preprocessing import StandardScaler

# Import PCA for dimensionality reduction
from sklearn.decomposition import PCA


# Separate the Wav2Vec2 features from acoustic features
X_train_w2v = X_train_raw[:, :1024]

# Separate the test Wav2Vec2 features
X_test_w2v = X_test_raw[:, :1024]

# Separate the acoustic features
X_train_acoustic = X_train_raw[:, 1024:]

# Separate the test acoustic features
X_test_acoustic = X_test_raw[:, 1024:]


# Create a scaler for Wav2Vec2 features
w2v_scaler = StandardScaler()

# Fit the scaler only on training Wav2Vec2 features
X_train_w2v_scaled = w2v_scaler.fit_transform(
    X_train_w2v
)

# Transform test Wav2Vec2 features using the same scaler
X_test_w2v_scaled = w2v_scaler.transform(
    X_test_w2v
)


# Create PCA retaining 95% of the variance
w2v_pca = PCA(
    n_components=0.95,
    random_state=42
)

# Fit PCA using only training data
X_train_w2v_pca = w2v_pca.fit_transform(
    X_train_w2v_scaled
)

# Transform the test data
X_test_w2v_pca = w2v_pca.transform(
    X_test_w2v_scaled
)


# Scale the acoustic features
acoustic_scaler = StandardScaler()

# Fit acoustic scaler on training data
X_train_acoustic_scaled = acoustic_scaler.fit_transform(
    X_train_acoustic
)

# Transform test acoustic features
X_test_acoustic_scaled = acoustic_scaler.transform(
    X_test_acoustic
)


# Combine reduced Wav2Vec2 features with acoustic features
X_train = np.hstack([
    X_train_w2v_pca,
    X_train_acoustic_scaled
])

# Combine test features
X_test = np.hstack([
    X_test_w2v_pca,
    X_test_acoustic_scaled
])


# Print PCA information
print(
    "Original Wav2Vec2 dimensions:",
    X_train_w2v.shape[1]
)

print(
    "Reduced Wav2Vec2 dimensions:",
    X_train_w2v_pca.shape[1]
)

print(
    "Explained variance:",
    round(
        w2v_pca.explained_variance_ratio_.sum(),
        4
    )
)

# Print final model feature dimensions
print(
    "Final train shape:",
    X_train.shape
)

print(
    "Final test shape:",
    X_test.shape
)
```

```python
# Import KFold for cross-validation
from sklearn.model_selection import KFold

# Import regression evaluation metric
from sklearn.metrics import mean_squared_error

# Import Pearson correlation
from scipy.stats import pearsonr

# Import Ridge regression
from sklearn.linear_model import Ridge

# Import Support Vector Regression
from sklearn.svm import SVR

# Import Random Forest
from sklearn.ensemble import RandomForestRegressor

# Import XGBoost
from xgboost import XGBRegressor

# Import LightGBM
from lightgbm import LGBMRegressor

# Import CatBoost
from catboost import CatBoostRegressor

# Import NumPy
import numpy as np


# Create 10-fold cross-validation
kf = KFold(
    n_splits=10,
    shuffle=True,
    random_state=42
)


# Create OOF predictions for XGBoost
oof_xgb = np.zeros(len(X_train))

# Create OOF predictions for LightGBM
oof_lgb = np.zeros(len(X_train))

# Create OOF predictions for CatBoost
oof_cat = np.zeros(len(X_train))

# Create OOF predictions for Random Forest
oof_rf = np.zeros(len(X_train))

# Create OOF predictions for Ridge
oof_ridge = np.zeros(len(X_train))

# Create OOF predictions for SVR
oof_svr = np.zeros(len(X_train))


# Create test prediction arrays
test_xgb = np.zeros(len(X_test))
test_lgb = np.zeros(len(X_test))
test_cat = np.zeros(len(X_test))
test_rf = np.zeros(len(X_test))
test_ridge = np.zeros(len(X_test))
test_svr = np.zeros(len(X_test))


# Start 10-fold cross-validation
for fold, (train_idx, valid_idx) in enumerate(
    kf.split(X_train),
    start=1
):

    # Print the current fold
    print(f"\n========== Fold {fold}/10 ==========")

    # Get training features for this fold
    X_tr = X_train[train_idx]

    # Get validation features for this fold
    X_val = X_train[valid_idx]

    # Get training targets for this fold
    y_tr = y[train_idx]

    # Get validation targets for this fold
    y_val = y[valid_idx]


    # -----------------------------
    # XGBoost
    # -----------------------------

    # Create the XGBoost model
    xgb_model = XGBRegressor(
        n_estimators=1000,
        learning_rate=0.025,
        max_depth=5,
        min_child_weight=5,
        subsample=0.85,
        colsample_bytree=0.85,
        reg_alpha=0.1,
        reg_lambda=2.0,
        objective="reg:squarederror",
        random_state=42,
        n_jobs=-1
    )

    # Train XGBoost
    xgb_model.fit(
        X_tr,
        y_tr
    )

    # Generate validation predictions
    oof_xgb[valid_idx] = xgb_model.predict(X_val)

    # Generate test predictions and average across folds
    test_xgb += xgb_model.predict(X_test) / 10


    # -----------------------------
    # LightGBM
    # -----------------------------

    # Create the LightGBM model
    lgb_model = LGBMRegressor(
        n_estimators=1000,
        learning_rate=0.025,
        max_depth=6,
        num_leaves=31,
        subsample=0.85,
        colsample_bytree=0.85,
        reg_alpha=0.1,
        reg_lambda=2.0,
        random_state=42,
        verbosity=-1,
        n_jobs=-1
    )

    # Train LightGBM
    lgb_model.fit(
        X_tr,
        y_tr
    )

    # Generate validation predictions
    oof_lgb[valid_idx] = lgb_model.predict(X_val)

    # Generate test predictions
    test_lgb += lgb_model.predict(X_test) / 10


    # -----------------------------
    # CatBoost
    # -----------------------------

    # Create the CatBoost model
    cat_model = CatBoostRegressor(
        iterations=1000,
        depth=6,
        learning_rate=0.03,
        loss_function="RMSE",
        verbose=False,
        random_seed=42
    )

    # Train CatBoost
    cat_model.fit(
        X_tr,
        y_tr
    )

    # Generate validation predictions
    oof_cat[valid_idx] = cat_model.predict(X_val)

    # Generate test predictions
    test_cat += cat_model.predict(X_test) / 10


    # -----------------------------
    # Random Forest
    # -----------------------------

    # Create the Random Forest model
    rf_model = RandomForestRegressor(
        n_estimators=500,
        max_features=0.75,
        min_samples_leaf=2,
        random_state=42,
        n_jobs=-1
    )

    # Train Random Forest
    rf_model.fit(
        X_tr,
        y_tr
    )

    # Generate validation predictions
    oof_rf[valid_idx] = rf_model.predict(X_val)

    # Generate test predictions
    test_rf += rf_model.predict(X_test) / 10


    # -----------------------------
    # Ridge
    # -----------------------------

    # Create Ridge regression
    ridge_model = Ridge(
        alpha=10.0
    )

    # Train Ridge
    ridge_model.fit(
        X_tr,
        y_tr
    )

    # Generate validation predictions
    oof_ridge[valid_idx] = ridge_model.predict(X_val)

    # Generate test predictions
    test_ridge += ridge_model.predict(X_test) / 10


    # -----------------------------
    # SVR
    # -----------------------------

    # Create Support Vector Regression
    svr_model = SVR(
        C=10.0,
        epsilon=0.05,
        gamma="scale"
    )

    # Train SVR
    svr_model.fit(
        X_tr,
        y_tr
    )

    # Generate validation predictions
    oof_svr[valid_idx] = svr_model.predict(X_val)

    # Generate test predictions
    test_svr += svr_model.predict(X_test) / 10


    # Calculate fold RMSE using the current sklearn-compatible method
    fold_rmse = np.sqrt(
        mean_squared_error(
            y_val,
            oof_xgb[valid_idx]
        )
    )

    # Print the fold RMSE
    print(
        "XGBoost fold RMSE:",
        round(fold_rmse, 4)
    )


# Print completion message
print("\n10-fold base-model training completed.")
```

```python
# Define a helper function to calculate RMSE and Pearson correlation
def evaluate_model(name, y_true, predictions):

    # Calculate RMSE using the sklearn-compatible formula
    rmse = np.sqrt(
        mean_squared_error(
            y_true,
            predictions
        )
    )

    # Calculate Pearson correlation
    pearson = pearsonr(
        y_true,
        predictions
    )[0]

    # Print the model name
    print(f"{name}")

    # Print RMSE
    print(f"  RMSE: {rmse:.4f}")

    # Print Pearson correlation
    print(f"  Pearson: {pearson:.4f}")

    # Return both metrics
    return rmse, pearson


# Evaluate XGBoost OOF predictions
xgb_score = evaluate_model(
    "XGBoost",
    y,
    oof_xgb
)

# Evaluate LightGBM OOF predictions
lgb_score = evaluate_model(
    "LightGBM",
    y,
    oof_lgb
)

# Evaluate CatBoost OOF predictions
cat_score = evaluate_model(
    "CatBoost",
    y,
    oof_cat
)

# Evaluate Random Forest OOF predictions
rf_score = evaluate_model(
    "Random Forest",
    y,
    oof_rf
)

# Evaluate Ridge OOF predictions
ridge_score = evaluate_model(
    "Ridge",
    y,
    oof_ridge
)

# Evaluate SVR OOF predictions
svr_score = evaluate_model(
    "SVR",
    y,
    oof_svr
)
```

```python
# Combine all out-of-fold base-model predictions into one matrix
meta_train = np.column_stack([
    oof_xgb,
    oof_lgb,
    oof_cat,
    oof_rf,
    oof_ridge,
    oof_svr
])

# Combine all test predictions into one matrix
meta_test = np.column_stack([
    test_xgb,
    test_lgb,
    test_cat,
    test_rf,
    test_ridge,
    test_svr
])

# Print the meta-feature shapes
print("Meta train shape:", meta_train.shape)
print("Meta test shape:", meta_test.shape)


# Create the Ridge meta-model
meta_model = Ridge(
    alpha=1.0
)

# Train the meta-model using ONLY out-of-fold predictions
meta_model.fit(
    meta_train,
    y
)

# Generate final OOF predictions from the stacking model
stacked_oof = meta_model.predict(
    meta_train
)

# Calculate stacking RMSE
stacked_rmse = np.sqrt(
    mean_squared_error(
        y,
        stacked_oof
    )
)

# Calculate stacking Pearson correlation
stacked_pearson = pearsonr(
    y,
    stacked_oof
)[0]

# Print stacking performance
print("\n===== STACKING RESULTS =====")
print(
    "Stacking RMSE:",
    round(stacked_rmse, 4)
)
print(
    "Stacking Pearson:",
    round(stacked_pearson, 4)
)


# Display the learned meta-model weights
print("\nMeta-model coefficients:")

# Define model names
model_names = [
    "XGBoost",
    "LightGBM",
    "CatBoost",
    "Random Forest",
    "Ridge",
    "SVR"
]

# Print each model's learned coefficient
for name, coefficient in zip(
    model_names,
    meta_model.coef_
):
    print(
        f"{name}: {coefficient:.4f}"
    )
```

```python
# ============================================================
# V4 — Linguistic / Grammar Feature Extraction
# ============================================================

import re
import numpy as np
import pandas as pd

# Common English filler words
FILLER_WORDS = {
    "um", "uh", "erm", "hmm", "ah",
    "like", "actually", "basically", "literally"
}


def extract_linguistic_features(text, duration):
    """
    Extract linguistic, fluency and vocabulary features
    from a Whisper transcript.
    """

    # Make sure text is a valid string
    text = str(text).lower().strip()

    # Extract alphabetic words
    words = re.findall(
        r"\b[a-zA-Z']+\b",
        text
    )

    word_count = len(words)

    # Unique vocabulary
    unique_words = len(set(words))

    # Type-token ratio
    type_token_ratio = (
        unique_words / word_count
        if word_count > 0 else 0
    )

    # Split into sentences
    sentences = re.split(
        r"[.!?]+",
        text
    )

    sentences = [
        s.strip()
        for s in sentences
        if s.strip()
    ]

    sentence_count = len(sentences)

    # Sentence lengths
    sentence_lengths = []

    for sentence in sentences:

        sentence_words = re.findall(
            r"\b[a-zA-Z']+\b",
            sentence
        )

        sentence_lengths.append(
            len(sentence_words)
        )

    if sentence_lengths:

        avg_sentence_length = np.mean(
            sentence_lengths
        )

        max_sentence_length = np.max(
            sentence_lengths
        )

        sentence_length_std = np.std(
            sentence_lengths
        )

    else:

        avg_sentence_length = 0
        max_sentence_length = 0
        sentence_length_std = 0

    # Protect against zero duration
    duration = max(float(duration), 1.0)

    # Speaking rate
    words_per_second = (
        word_count / duration
    )

    words_per_minute = (
        words_per_second * 60
    )

    # --------------------------------------------------------
    # Filler words
    # --------------------------------------------------------

    filler_count = 0

    for filler in FILLER_WORDS:

        filler_count += len(
            re.findall(
                r"\b" + re.escape(filler) + r"\b",
                text
            )
        )

    filler_ratio = (
        filler_count / word_count
        if word_count > 0 else 0
    )

    # --------------------------------------------------------
    # Repeated consecutive words
    # --------------------------------------------------------

    repeated_words = 0

    for i in range(1, len(words)):

        if words[i] == words[i - 1]:

            repeated_words += 1

    repeated_word_ratio = (
        repeated_words / word_count
        if word_count > 0 else 0
    )

    # --------------------------------------------------------
    # Word characteristics
    # --------------------------------------------------------

    if word_count > 0:

        avg_word_length = np.mean([
            len(word)
            for word in words
        ])

        word_length_std = np.std([
            len(word)
            for word in words
        ])

    else:

        avg_word_length = 0
        word_length_std = 0

    # Long vocabulary
    long_words = sum(
        len(word) >= 7
        for word in words
    )

    long_word_ratio = (
        long_words / word_count
        if word_count > 0 else 0
    )

    # --------------------------------------------------------
    # Punctuation / structure
    # --------------------------------------------------------

    comma_count = text.count(",")

    question_count = text.count("?")

    exclamation_count = text.count("!")

    # Character count
    character_count = len(text)

    # Average characters per word
    avg_chars_per_word = (
        character_count / word_count
        if word_count > 0 else 0
    )

    # --------------------------------------------------------
    # Return feature dictionary
    # --------------------------------------------------------

    return {

        "word_count":
            word_count,

        "unique_word_count":
            unique_words,

        "type_token_ratio":
            type_token_ratio,

        "sentence_count":
            sentence_count,

        "avg_sentence_length":
            avg_sentence_length,

        "max_sentence_length":
            max_sentence_length,

        "sentence_length_std":
            sentence_length_std,

        "words_per_second":
            words_per_second,

        "words_per_minute":
            words_per_minute,

        "filler_count":
            filler_count,

        "filler_ratio":
            filler_ratio,

        "repeated_words":
            repeated_words,

        "repeated_word_ratio":
            repeated_word_ratio,

        "avg_word_length":
            avg_word_length,

        "word_length_std":
            word_length_std,

        "long_word_ratio":
            long_word_ratio,

        "comma_count":
            comma_count,

        "question_count":
            question_count,

        "exclamation_count":
            exclamation_count,

        "character_count":
            character_count,

        "avg_chars_per_word":
            avg_chars_per_word
    }
```

```python
# ============================================================
# V4 — Reliable Acoustic Feature Extraction
# ============================================================

import os
import numpy as np
import pandas as pd
import librosa
from tqdm.auto import tqdm


def extract_acoustic_features_v4(audio_path):
    """
    Extract 25 acoustic/prosodic features from one WAV file.
    """

    # Load audio at 16 kHz
    y, sr = librosa.load(
        audio_path,
        sr=16000,
        mono=True
    )

    # Handle empty audio
    if len(y) == 0:
        return np.zeros(25, dtype=np.float32)

    # Duration
    duration = len(y) / sr

    # RMS energy
    rms = librosa.feature.rms(y=y)[0]

    # Zero crossing rate
    zcr = librosa.feature.zero_crossing_rate(y)[0]

    # Spectral centroid
    centroid = librosa.feature.spectral_centroid(
        y=y,
        sr=sr
    )[0]

    # Spectral bandwidth
    bandwidth = librosa.feature.spectral_bandwidth(
        y=y,
        sr=sr
    )[0]

    # Spectral rolloff
    rolloff = librosa.feature.spectral_rolloff(
        y=y,
        sr=sr
    )[0]

    # Pitch
    f0 = librosa.yin(
        y,
        fmin=70,
        fmax=400,
        sr=sr
    )

    # Valid pitch values
    valid_f0 = f0[np.isfinite(f0)]

    # Onset strength
    onset = librosa.onset.onset_strength(
        y=y,
        sr=sr
    )

    # Energy change
    energy_change = np.diff(rms)

    # Silence ratio
    silence_ratio = np.mean(
        rms < (np.max(rms) * 0.05)
    )

    # Speech ratio
    speech_ratio = 1.0 - silence_ratio

    # Onset rate
    onset_count = len(
        librosa.onset.onset_detect(
            y=y,
            sr=sr
        )
    )

    onset_rate = (
        onset_count / max(duration, 1e-6)
    )

    # Pitch statistics
    if len(valid_f0) > 0:

        pitch_mean = np.mean(valid_f0)
        pitch_std = np.std(valid_f0)
        pitch_min = np.min(valid_f0)
        pitch_max = np.max(valid_f0)

    else:

        pitch_mean = 0.0
        pitch_std = 0.0
        pitch_min = 0.0
        pitch_max = 0.0

    # Exactly 25 features
    features = [

        duration,

        np.mean(rms),
        np.std(rms),
        np.min(rms),
        np.max(rms),

        np.mean(zcr),
        np.std(zcr),

        np.mean(centroid),
        np.std(centroid),

        np.mean(bandwidth),
        np.std(bandwidth),

        np.mean(rolloff),
        np.std(rolloff),

        pitch_mean,
        pitch_std,
        pitch_min,
        pitch_max,

        np.mean(onset),
        np.std(onset),

        np.mean(energy_change)
            if len(energy_change) else 0.0,

        np.std(energy_change)
            if len(energy_change) else 0.0,

        silence_ratio,
        speech_ratio,

        onset_rate,

        len(valid_f0)
    ]

    return np.asarray(
        features,
        dtype=np.float32
    )


# ------------------------------------------------------------
# TRAIN ACOUSTIC FEATURES
# ------------------------------------------------------------

train_acoustic = []

for filename in tqdm(
    train_df["filename"].tolist(),
    desc="Train acoustic features"
):

    audio_path = os.path.join(
        TRAIN_DIR,
        filename
    )

    train_acoustic.append(
        extract_acoustic_features_v4(
            audio_path
        )
    )


# Convert list of vectors → 2D matrix
train_acoustic = np.vstack(
    train_acoustic
)


# ------------------------------------------------------------
# TEST ACOUSTIC FEATURES
# ------------------------------------------------------------

test_acoustic = []

for filename in tqdm(
    test_df["filename"].tolist(),
    desc="Test acoustic features"
):

    audio_path = os.path.join(
        TEST_DIR,
        filename
    )

    test_acoustic.append(
        extract_acoustic_features_v4(
            audio_path
        )
    )


# Convert list of vectors → 2D matrix
test_acoustic = np.vstack(
    test_acoustic
)


# ------------------------------------------------------------
# CHECK
# ------------------------------------------------------------

print(
    "Train acoustic shape:",
    train_acoustic.shape
)

print(
    "Test acoustic shape:",
    test_acoustic.shape
)

print(
    "Train NaNs:",
    np.isnan(train_acoustic).sum()
)

print(
    "Test NaNs:",
    np.isnan(test_acoustic).sum()
)
```

```python
# ============================================================
# V4 — Load and combine all feature groups
# ============================================================

import numpy as np
import pandas as pd

# ------------------------------------------------------------
# 1. Load Wav2Vec2 embeddings
# ------------------------------------------------------------

train_embeddings = np.load(
    "/kaggle/working/train_wav2vec.npy"
)

test_embeddings = np.load(
    "/kaggle/working/test_wav2vec.npy"
)

print("Wav2Vec2 train:", train_embeddings.shape)
print("Wav2Vec2 test :", test_embeddings.shape)


# ------------------------------------------------------------
# 2. Load acoustic features
# ------------------------------------------------------------

train_acoustic = np.load(
    "/kaggle/working/train_acoustic_v4.npy"
)

test_acoustic = np.load(
    "/kaggle/working/test_acoustic_v4.npy"
)

print("Acoustic train:", train_acoustic.shape)
print("Acoustic test :", test_acoustic.shape)


# ------------------------------------------------------------
# 3. Load linguistic features
# ------------------------------------------------------------

ling_train = pd.read_csv(
    "/kaggle/working/train_linguistic_features.csv"
)

ling_test = pd.read_csv(
    "/kaggle/working/test_linguistic_features.csv"
)

ling_train_array = ling_train.to_numpy(
    dtype=np.float32
)

ling_test_array = ling_test.to_numpy(
    dtype=np.float32
)

print("Linguistic train:", ling_train_array.shape)
print("Linguistic test :", ling_test_array.shape)


# ------------------------------------------------------------
# 4. Combine Wav2Vec2 + acoustic + linguistic
# ------------------------------------------------------------

X_train_v4_raw = np.hstack([
    train_embeddings,
    train_acoustic,
    ling_train_array
])

X_test_v4_raw = np.hstack([
    test_embeddings,
    test_acoustic,
    ling_test_array
])


# ------------------------------------------------------------
# 5. Final shape check
# ------------------------------------------------------------

print("\n========================================")
print("V4 FEATURE MATRIX")
print("========================================")

print(
    "Train:",
    X_train_v4_raw.shape
)

print(
    "Test :",
    X_test_v4_raw.shape
)

print(
    "Train NaNs:",
    np.isnan(X_train_v4_raw).sum()
)

print(
    "Test NaNs:",
    np.isnan(X_test_v4_raw).sum()
)
```

```python
# ============================================================
# V4 — PCA for Wav2Vec2 + scaling for acoustic/linguistic
# ============================================================

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# ------------------------------------------------------------
# Separate feature groups
# ------------------------------------------------------------

# Wav2Vec2 = first 1024 columns
wav_train = X_train_v4_raw[:, :1024]
wav_test = X_test_v4_raw[:, :1024]

# Acoustic = next 25 columns
acoustic_train_v4 = X_train_v4_raw[:, 1024:1049]
acoustic_test_v4 = X_test_v4_raw[:, 1024:1049]

# Linguistic = final 21 columns
ling_train_v4 = X_train_v4_raw[:, 1049:]
ling_test_v4 = X_test_v4_raw[:, 1049:]


print("Wav2Vec2:", wav_train.shape)
print("Acoustic:", acoustic_train_v4.shape)
print("Linguistic:", ling_train_v4.shape)


# ------------------------------------------------------------
# Scale Wav2Vec2
# ------------------------------------------------------------

wav_scaler = StandardScaler()

wav_train_scaled = wav_scaler.fit_transform(
    wav_train
)

wav_test_scaled = wav_scaler.transform(
    wav_test
)


# ------------------------------------------------------------
# PCA — retain 95% variance
# ------------------------------------------------------------

pca = PCA(
    n_components=0.95,
    random_state=42
)

wav_train_pca = pca.fit_transform(
    wav_train_scaled
)

wav_test_pca = pca.transform(
    wav_test_scaled
)


print("\nWav2Vec2 original dimensions:", 1024)
print(
    "PCA dimensions:",
    wav_train_pca.shape[1]
)

print(
    "Explained variance:",
    round(
        pca.explained_variance_ratio_.sum(),
        4
    )
)
```

```python
# ============================================================
# Scale acoustic + linguistic features
# ============================================================

# Combine acoustic and linguistic features
small_train = np.hstack([
    acoustic_train_v4,
    ling_train_v4
])

small_test = np.hstack([
    acoustic_test_v4,
    ling_test_v4
])

print(
    "Acoustic + linguistic train:",
    small_train.shape
)

print(
    "Acoustic + linguistic test:",
    small_test.shape
)


# Standardize
small_scaler = StandardScaler()

small_train_scaled = small_scaler.fit_transform(
    small_train
)

small_test_scaled = small_scaler.transform(
    small_test
)


print("Scaling completed.")
```

```python
# ============================================================
# Final V4 feature matrix
# ============================================================

X_train_v4 = np.hstack([
    wav_train_pca,
    small_train_scaled
])

X_test_v4 = np.hstack([
    wav_test_pca,
    small_test_scaled
])


print(
    "Final V4 train shape:",
    X_train_v4.shape
)

print(
    "Final V4 test shape:",
    X_test_v4.shape
)

print(
    "NaNs train:",
    np.isnan(X_train_v4).sum()
)

print(
    "NaNs test:",
    np.isnan(X_test_v4).sum()
)
```

```python
# ============================================================
# V4 — 10-Fold Ensemble + OOF Stacking
# ============================================================

import numpy as np
import pandas as pd

from sklearn.model_selection import KFold
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
from sklearn.ensemble import RandomForestRegressor
from sklearn.svm import SVR

from xgboost import XGBRegressor
from lightgbm import LGBMRegressor
from catboost import CatBoostRegressor

from scipy.stats import pearsonr
from tqdm.auto import tqdm


# ------------------------------------------------------------
# Target
# ------------------------------------------------------------

y = train_df["label"].values.astype(np.float32)

print("Target shape:", y.shape)


# ------------------------------------------------------------
# 10-fold cross-validation
# ------------------------------------------------------------

kf = KFold(
    n_splits=10,
    shuffle=True,
    random_state=42
)


# ------------------------------------------------------------
# Model factory
# ------------------------------------------------------------

def create_models():

    return {

        "XGB": XGBRegressor(
            n_estimators=1000,
            learning_rate=0.025,
            max_depth=5,
            min_child_weight=5,
            subsample=0.85,
            colsample_bytree=0.85,
            reg_alpha=0.1,
            reg_lambda=2,
            objective="reg:squarederror",
            random_state=42,
            n_jobs=-1
        ),

        "LGBM": LGBMRegressor(
            n_estimators=1000,
            learning_rate=0.025,
            max_depth=6,
            num_leaves=31,
            subsample=0.85,
            colsample_bytree=0.85,
            reg_alpha=0.1,
            reg_lambda=2,
            random_state=42,
            verbosity=-1,
            n_jobs=-1
        ),

        "CAT": CatBoostRegressor(
            iterations=1000,
            depth=6,
            learning_rate=0.03,
            loss_function="RMSE",
            verbose=False,
            random_seed=42
        ),

        "RF": RandomForestRegressor(
            n_estimators=500,
            max_features=0.75,
            min_samples_leaf=2,
            random_state=42,
            n_jobs=-1
        ),

        "Ridge": Ridge(
            alpha=10.0
        ),

        "SVR": SVR(
            C=10,
            epsilon=0.05,
            gamma="scale"
        )
    }


# ------------------------------------------------------------
# Storage
# ------------------------------------------------------------

model_names = [
    "XGB",
    "LGBM",
    "CAT",
    "RF",
    "Ridge",
    "SVR"
]

oof_predictions = {
    name: np.zeros(len(y))
    for name in model_names
}

test_predictions = {
    name: np.zeros(len(X_test_v4))
    for name in model_names
}


# ------------------------------------------------------------
# Cross-validation
# ------------------------------------------------------------

for fold, (train_idx, valid_idx) in enumerate(
    kf.split(X_train_v4),
    start=1
):

    print(
        f"\n========== Fold {fold}/10 =========="
    )

    X_tr = X_train_v4[train_idx]
    X_val = X_train_v4[valid_idx]

    y_tr = y[train_idx]
    y_val = y[valid_idx]

    models = create_models()

    for name, model in models.items():

        print(
            f"Training {name}..."
        )

        # Train
        model.fit(
            X_tr,
            y_tr
        )

        # Validation prediction
        val_pred = model.predict(
            X_val
        )

        # Test prediction
        test_pred = model.predict(
            X_test_v4
        )

        # Store OOF prediction
        oof_predictions[name][valid_idx] = (
            val_pred
        )

        # Average test predictions across folds
        test_predictions[name] += (
            test_pred / 10
        )


print("\nAll folds completed.")
```

```python
# ============================================================
# Evaluate individual V4 models
# ============================================================

results = []

for name in model_names:

    rmse = np.sqrt(
        mean_squared_error(
            y,
            oof_predictions[name]
        )
    )

    pearson = pearsonr(
        y,
        oof_predictions[name]
    )[0]

    results.append({
        "Model": name,
        "RMSE": rmse,
        "Pearson": pearson
    })


results_df = pd.DataFrame(
    results
).sort_values(
    "RMSE"
)

display(results_df)
```

```python
# ============================================================
# V4 — Ridge Stacking
# ============================================================

# OOF predictions become meta features
meta_train = np.column_stack([
    oof_predictions[name]
    for name in model_names
])

# Test predictions become meta-test features
meta_test = np.column_stack([
    test_predictions[name]
    for name in model_names
])

print("Meta train:", meta_train.shape)
print("Meta test :", meta_test.shape)


# Meta learner
meta_model = Ridge(
    alpha=1.0
)

meta_model.fit(
    meta_train,
    y
)


# OOF stacked prediction
stacked_oof = meta_model.predict(
    meta_train
)


# Evaluate stack
stack_rmse = np.sqrt(
    mean_squared_error(
        y,
        stacked_oof
    )
)

stack_pearson = pearsonr(
    y,
    stacked_oof
)[0]


print("\n==============================")
print("V4 STACKING RESULTS")
print("==============================")

print(
    "OOF RMSE:",
    round(stack_rmse, 4)
)

print(
    "OOF Pearson:",
    round(stack_pearson, 4)
)

print(
    "\nV3 benchmark:"
)

print(
    "RMSE: 0.6685"
)

print(
    "Pearson: 0.8417"
)
```

```python
# ============================================================
# V4 — Generate final test predictions
# ============================================================

# Meta-model prediction on test
v4_test_pred = meta_model.predict(
    meta_test
)

# Keep scores within the competition range
v4_test_pred = np.clip(
    v4_test_pred,
    0,
    5
)

print("Prediction count:", len(v4_test_pred))
print("Minimum:", v4_test_pred.min())
print("Maximum:", v4_test_pred.max())
print("Mean:", v4_test_pred.mean())
```
